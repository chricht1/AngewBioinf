```python
from Bio.Seq import Seq

mySeq = Seq("ATGACGGCTACGTAG")
mySeq[3:7]

print(mySeq[3:7])

str(mySeq[3:7])

dna1 = Seq("ACGT")
dna2 = Seq("GAAC")
print(dna1 + dna2)


prot = Seq("RPRNAK")

print( dna1 + prot )


mySeq = Seq("ATGACGGCTACGTAG")
print( mySeq.reverse_complement() )

print( mySeq.reverse_complement().transcribe() )

print( mySeq.translate() )
# output: MTAT* 
# * is stop symbol, indicates stop codon
```
