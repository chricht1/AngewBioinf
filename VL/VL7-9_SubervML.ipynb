### Entscheidungsbäume

- Binärbaum mit Entscheidungsregeln

- klassisch für Klassifik. Anwendung

- auch prinzipiell Anwendung für Regression, dann stehen in Blättern konkrete Werte

Wie Bestimmung der Entscheidungsregeln?
- Homogenitätsmaße: z. B.: Entropieminimierung

- so lange Aufteilung bis Max Tiefe erreicht, oder Homogenitätszielwert erreicht
  - Maximaltiefe u. geringerer Homogenitätszielwert reduziert Overfitting

Anwendung auf Daten:
- Overfitting (Testperformanz deutl. schlechter als Trainingsperformanz)

### SVM

- teile Eingaberaum mittels Hyperebenen in Regionen

- Hyperebene durch Normalenvektor u. Versch.-Vektor gebildet

  - Maximierung d. Margin zwischen Hyperebene u. des jew. nächsten Punktes(Support-Vektoren) der Klasse zur Ebene für alle Klassen

- 

- eigentl. numerische Opt. über duales Problem

#### SVM - Kernel
- transform. Daten über Fkt. phi() in höherdim. Raum

- radiale Basisfunktion (RBF) ist beliebte Wahl für Kernel-Fkt.
- 'linear' ist kein Kernel

- Kernel-Trick:
  - Proj. u. Skalarprod. muss nicht explizit ausgerechnet bzw. durchführt werden sondern kann direkt durch Fkt. ausgerechnet werden
- SVM mit RBF Transformation nur durch Kernel-Trick möglich

Anwendung: 
- fit dauert deutl. länger, wegen opt. duales Problem
