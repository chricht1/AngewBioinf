Regex Quantifikatoren:

![Quantifikatoren](./Images/re_quant.png)

```python
# reguläre Ausdrücke:
import re # Paket für Regexes
s = "ACATGATAGGCGTATA"
rex = r"((TA){2}|(TA(C|G)+))" # Escapes nicht aufgeloest
```

#### nützliche Methoden:

```python
# search() - sucht nach 1. Vorkommen

matchObject = re.search(rex,s,re.I) # Klasse re.MatchObject
# weitere Flages/Modifier, z.B. re.I für "ignore case" (Groß- Kleinschr.)
if matchObject:
  print(matchObject)
  print(matchObject.group())
```

```python
# sub() - Modifikation des gematchten strings
# re.sub(rex, replace, stri)

# Bsp. progressive/rekursive Ersetzung:

# Zerlegung in 3er- Blöcke
line = "1234567890";
while True:
  line2 = re.sub(r"([0-9]+)([0-9]{3})($| )",r"\1 \2\3",line)
  # \i referenziert n-ten Match
  if line2 == line:
    break
  line = line2
  print(line)
```

```python
# findIter() - gibt Iterator über alle Vorkommen zurück
# Achtung: es werden keine überlappenden Vorkommen erkannt

print(re.finditer(rex,s))
print(iter(re.finditer(rex,s)))
print(next(iter(re.finditer(rex,s))))

for obj in re.finditer(rex,s):
  print("({0},{1})".format(obj.start(),obj.end()) )
```

```python
# progressive Ersetzung:

# Zerlegung in 3er- Blöcke
line = "1234567890";
while True:
  line2 = re.sub(r"([0-9]+)([0-9]{3})($| )",r"\1 \2\3",line)
  # \i referenziert n-ten Match
  if line2 == line:
    break
  line = line2
  print(line)
```
